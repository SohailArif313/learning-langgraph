```python
from langgraph.graph import StateGraph,START,END
from typing import TypedDict
```

```python
# define state
class BmiState(TypedDict):
    weight: float
    height:float
    verdict:str
    bmi: float
    status:bool
    age:int
    name:str
```

```python
def calculate_bmi(state: BmiState) -> BmiState:
    weight = state["weight"]
    height = state["height"]
    age = state["age"]

    bmi = weight / (height ** 2)

    if age < 16:
        state["status"] = False
    elif age <= 40:
        state["status"] = True
    else:
        state["status"] = False

    state["bmi"] = round(bmi, 2)

    if bmi < 18.5:
        state["verdict"] = "Underweight"
    elif bmi < 25:
        state["verdict"] = "Normal"
    elif bmi < 30:
        state["verdict"] = "Overweight"
    else:
        state["verdict"] = "Obese"

    if 18.5 <= bmi <= 25 and 16 <= age <= 40:
        print(
            f"Congratulations {state['name']}, you can join our gym\n"
            f"Your age: {state['age']}\n"
            f"BMI: {state['bmi']}\n"
            f"Status: {state['status']}"
        )

    return state
```

```python
graph = StateGraph(BmiState)
graph.add_node("bmi_calculator",calculate_bmi)

graph.add_edge(START,"bmi_calculator")
graph.add_edge("bmi_calculator",END)

workflow = graph.compile()
```

```python
age = int(input("Enter your age: "))
name = input("Enter your name: ").title()
print(name)
initial_state = {"weight":80,"height":1.70,"age":age,"name":name}
final_state = workflow.invoke(initial_state)
print(final_state)
```
